```python
import sys
sys.path.append('..')  # Add project root to path

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Import project modules
from src._01_setup import config_loader, checkpoint_manager

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Load Consolidated Data

Quick access to consolidated CSVs from the pipeline.

```python
# Load cluster assignments
df_assignments = pd.read_csv('../output/germany/01_data/assignments.csv')

print(f"✓ Loaded {len(df_assignments)} company assignments")
print(f"✓ Clusters: {df_assignments['cluster'].nunique()}")
print(f"\nFirst 5 companies:")
df_assignments.head()
```

```python
# Load cluster profiles
df_profiles = pd.read_csv('../output/germany/01_data/profiles.csv')

print("✓ Cluster Profiles:")
df_profiles
```

## 2. Basic Analysis

Quick exploration of cluster characteristics.

```python
# Cluster sizes
cluster_sizes = df_assignments['cluster'].value_counts().sort_index()

plt.figure(figsize=(10, 5))
cluster_sizes.plot(kind='bar', color='steelblue')
plt.title('Cluster Sizes', fontsize=14, fontweight='bold')
plt.xlabel('Cluster')
plt.ylabel('Number of Companies')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print(f"\n📊 Cluster Distribution:")
for cluster, count in cluster_sizes.items():
    print(f"  Cluster {cluster}: {count} companies ({count/len(df_assignments)*100:.1f}%)")
```

```python
# Cluster profiles heatmap
features = ['roa', 'roe', 'ebit_margin', 'gross_margin', 'fcf_margin', 
            'current_ratio', 'debt_to_equity', 'financial_leverage']

plt.figure(figsize=(12, 6))
sns.heatmap(df_profiles[features].T, annot=True, fmt='.2f', cmap='RdYlGn', 
            center=0, cbar_kws={'label': 'Value'})
plt.title('Cluster Profiles (Feature Means)', fontsize=14, fontweight='bold')
plt.xlabel('Cluster')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()
```
